```python
sns.histplot(df['2023 population'], bins = 10, kde = True)
plt.title("Distribution of 2023 Population")
```

```python
df.columns
```

```python
df.groupby("continent")["2023 population"].mean().sort_values(ascending=False).head(2)
```

```python
top_10_countries = df.sort_values(by='2023 population', ascending=False).head(15)
```

```python
top_10_countries
```

```python
sns.barplot(data = top_10_countries, x = '2023 population', y = 'country')
plt.title("Top 10 countries by Population in 2023")
plt.xlabel("Population")
plt.ylabel("Country")
plt.show()
```

```python
df.columns
```

```python
sns.scatterplot(x = '2023 population', y = 'density (km²)', data = df)
plt.title("Population and Density Scatter plot")
```

```python
df.groupby("continent")['area (km²)'].sum().sort_values(ascending=False)
```

```python
sns.violinplot(x="continent", y = 'density (km²)', data = df)
plt.title("Population Density Distribution Across Continents")
plt.xlabel("Continent")
plt.ylabel("Population Density")
plt.show()
```

```python
sns.boxplot(data = df, x = 'continent', y = 'density (km²)', showfliers=False)
plt.title("Population Density Distribution Acoss Continents (without outliers)")
```

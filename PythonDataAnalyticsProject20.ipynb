# Mean value per year

```python
df.head()
```

```python
mean_val_per_year = df.groupby("Year")["Value (Actual)"].mean()
mean_val_per_year_head = mean_val_per_year.head(20)
```

```python
sns.barplot(x = mean_val_per_year_head.index, y = mean_val_per_year_head.values)
plt.xlabel("Year")
plt.ylabel("Mean value")
plt.xticks(rotation = 75)
plt.show()
```

# Value over time chart

```python
df.head()
```

```python
df_sorted = df.sort_values(by = "Year")
```

```python
df_sorted
```

```python
plt.figure(figsize = (10,5))
plt.plot(df_sorted["Year"], df_sorted["Value (Actual)"], marker = 'o')
plt.title("Value over time")
plt.xlabel("Year")
plt.ylabel("Value actual")
plt.grid(True)
plt.show()
```

# Distribution of value by format

```python
sns.boxplot(x = "Format", y = "Value (Actual)", data = df.head(300))
plt.title("Distribution of value by format")
plt.xticks(rotation = 75)
plt.xlabel("Format")
plt.ylabel("Value actual")
plt.show()
```

# Relationship between columns

# Total value per year by metric

```python
year_metric_sum  = df.groupby(["Year", "Metric"])["Value (Actual)"].sum().unstack()
```

```python
year_metric_sum
```

```python
plt.figure(figsize = (15,20))
year_metric_sum.plot(kind="bar", stacked=True)
plt.xticks(rotation = 75)
plt.tight_layout()
plt.show()
```

# Average value per format over time

```python
format_year_avg = df.groupby(['Format', 'Year'])['Value (Actual)'].mean().unstack()
format_year_avg.T.plot(marker='o')  # transpose for proper x-axis
plt.title("Average value per format over time")
plt.xlabel("Year")
plt.ylabel("Average value")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left') 
plt.tight_layout()
plt.show()
```

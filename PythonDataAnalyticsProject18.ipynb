# Distribution of Publishing Year

# Number of books in each genre

```python
df1.columns
```

```python
df1["genre"].value_counts().plot(kind = "bar")
plt.xticks(rotation=75)
plt.xlabel("Genre")
plt.ylabel("Number of Books")
plt.title("Numbe of books in each genre")
plt.show()
```

# Author by average rating

```python
df1.head(1)
```

# Box plot for book ratings count for each genre

```python
df1.head(1)
```

```python
genre_by_rating_count = df1.groupby("genre")["Book_ratings_count"].sum()
```

```python
genre_by_rating_count
```

```python
sns.boxplot(data = df1, x = "genre", y = "Book_ratings_count")
plt.xlabel("Genre")
plt.ylabel("Rating count")
plt.title("Genre by book rating count")
plt.show()
```

# Total gross sales by each author

```python
tot_gross_sales_by_author = df1.groupby("Author")["gross sales"].sum().sort_values(ascending=False).head(20)
```

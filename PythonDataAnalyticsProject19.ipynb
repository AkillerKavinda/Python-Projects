# Scatter plot for price vs other features

```python
# Cell 13: Scatter Plots for Price vs Other Features
colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink', 'gray', 'olive', 'cyan', 'magenta']
color_index = 0

for col in df.columns[:-1]:  # Exclude 'price'
    plt.figure(figsize=(6, 4))
    plt.scatter(df[col], df["Price"], color=colors[color_index % len(colors)])
    plt.xlabel(col)
    plt.ylabel("Price")
    plt.title(f"Price vs {col}")
    plt.tight_layout()
    plt.show()
    color_index += 1
```
